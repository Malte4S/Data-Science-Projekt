```python
# Imports
import requests as r
import pandas as pd
import time as t
```

```python
# Relevant countries and Bidding Zones for Research Question 4 and 5
countries = ["pl", "cz", "de", "fi"]
bz = ["DE-LU", "PL", "CZ", "FI"]
```

```python
# First request to get hourly prouktion per country
url1 = "https://api.energy-charts.info/public_power"
```

```python
# Transform request into table and turn unix time into readable datetime
for c in countries:

    print(f"Requesting {c.upper()}")
    
    params1 = {
    "country": c,
    "start": "2015-01-01",
    "end": "2026-01-01",
    }

    response1 = r.get(url1, params=params1)
    data1 = response1.json()

    df1 = pd.DataFrame({
        "timestamp": pd.to_datetime(data1["unix_seconds"], unit="s")
    })

    for production in data1["production_types"]:
        df1[production["name"]] = production["data"]

    display(df1)

    df1.to_csv(f"PG_{c.upper()}.csv", index=False)
```

```python
# Second request to get the actually installed power-production types per country
url2 = "https://api.energy-charts.info/v2/installed_power"
```

```python
# Transform request into table and clean up column names
for c in countries:
    
    print(f"Requesting {c.upper()}")
    
    params2 = {
        "country" : c
        ,"time_step" : "yearly"
    }

    response2 = r.get(url2, params=params2)
    data2 = response2.json()

    df2 = pd.json_normalize(data2["data"])
    df2.columns = df2.columns.str.replace("values.", "", regex=False)

    display(df2)

    df2.to_csv(f"{c}.csv", index=False)
```

```python
# Thrid request to get the Bidding Zone data for each country
url3 = "https://api.energy-charts.info/v2/price"
```

```python
# Transform request into table and clean up column names
for c in bz:

    print(f"Requesting {c.upper()}")

    params3 = {
        "bzn" : c
        ,"start" : "2015-01-01"
        ,"end" : "2025-01-01"
    }

    response3 = r.get(url3, params=params3)
    data3 = response3.json()

    df3 = pd.json_normalize(data3["data"])

    df3.columns = df3.columns.str.replace("values.", "", regex=False)

    display(df3)

    df3.to_csv(f"{c}_BZ.csv", index=False)
```

```python
# In this case we want to get weather data for Finnland
# Transform request into table and add positional data before storing the loaction dataframe in a list of successful requests. If things go wrong they are instead saved in a failed list for later retry

df_set = []
failed_loaction = []

for l in locations_fi:

    param4 = {
        "latitude" : l[0]
        ,"longitude" : l[1]
        ,"start_date" : "2018-01-01"
        ,"end_date" : "2025-01-01"
        ,"daily" : ""
        "temperature_2m_max,precipitation_sum,shortwave_radiation_sum,wind_speed_10m_mean,soil_moisture_28_to_100cm_mean"
    }

    print(f"Requesting Location {l}")

    try:

        response = r.get(url=url4, params=param4)
        print(response)

        data = response.json()
        df_temp = pd.DataFrame(data["daily"])

        # Add coordinates to each location data
        df_temp["latitude"] = l[0]
        df_temp["longitude"] = l[1]

        # Add location dataframe to dataset
        df_set.append(df_temp)

    except: 

        failed_loaction.append(l)

    t.sleep(20)
```

```python
# Get failed position for later retry
print(failed_loaction)
```

```python
# Concatinate all successful location dataframes into one
df_complete = pd.concat(
    df_set,
    ignore_index=True
)
```
